Even linear models like Logistic Regression are less likely to suffer from unstable coefficient estimates due to correlated predictors.

```python
#Class Imbalance
df["Credit_Score"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.show()
```

This chart breaks down the exact percentages of people in the dataset belonging to each Credit Score group (Standard, Poor, and Good). If you pick a random person from this dataset, there is a better-than-average chance they have an ordinary, everyday credit score. However, a person is significantly more likely to have a bad credit score than a great one.

- Because more than half the dataset is "Standard," a completely uneducated Machine Learning model could simply guess "Standard" for every single person and automatically achieve over 53% accuracy. You cannot use regular Accuracy as your metric to judge if your model is good. You must look at metrics like Macro F1-Score, Precision, and Recall to make sure the model is actually learning how to spot the rare "Good" credit scores.
- In a real banking or lending environment, misclassifying a borrower is highly costly. If you accidentally label a Poor borrower as Standard/Good, you risk high loan defaults. If you miss a Good borrower because the model favors the dominant classes, you lose excellent business to a competitor. The Action Item: Your model needs to be finely tuned to pick up the specific behavioral signatures of that small 17.8% "Good" group—such as the low payment delays we noticed in the previous charts

Fix
- Resampling Techniques: Oversampling: Synthetically generate more examples of the "Good" class (using techniques like SMOTE). Undersampling: Carefully reduce the number of "Standard" examples so the model doesn't get overwhelmed by them.
- Class Weights: Tell your Machine Learning algorithm (like a Random Forest or XGBoost) to penalize errors on the "Good" class much more heavily than errors on the "Standard" class. This forces the model to pay closer attention to the minority group.
