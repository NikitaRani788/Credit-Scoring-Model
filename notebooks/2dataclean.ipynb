Credit_History_Age

This column is not suitable as text for machine learning. Instead, convert it into total months.

```python
#Create a conversion function
import re

def convert_credit_history(value):
    if pd.isna(value):
        return np.nan

    match = re.match(r'(\d+)\s+Years?\s+and\s+(\d+)\s+Months?', str(value))

    if match:
        years = int(match.group(1))
        months = int(match.group(2))
        return years * 12 + months

    return np.nan
```

```python
#Apply function
df_clean["Credit_History_Age"] = (
    df_clean["Credit_History_Age"]
    .apply(convert_credit_history)
)
```

```python
#Fill missing values
df_clean["Credit_History_Age"].fillna(
    df_clean["Credit_History_Age"].median(),
    inplace=True
)
```

```python
#Detect outliers using IQR
def outliers(col):

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    return df_clean[(df_clean[col] < lower) | (df_clean[col] > upper)]
```

```python
outliers("Annual_Income")
```
